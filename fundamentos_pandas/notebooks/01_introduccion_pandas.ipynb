### Series

```python
serie_ventas = pd.Series(
    [120, 85, 200, 45],
    index=["enero", "febrero", "marzo", "abril"],
    name="ventas"
)

serie_ventas
```

### DataFrames

#### Crear Dataframe desde un diccionario

```python
df = pd.DataFrame({
    "producto": ["Laptop", "Monitor", "Teclado"],
    "unidades": [3, 7, 12],
    "precio": [1200, 320, 95]
})

df
```

#### Desde una lista de diccionarios (JSONs)

```python
df2 = pd.DataFrame(
    [
        {"producto": "Laptop", "unidades": 3, "precio": 1200},
        {"producto": "Monitor", "unidades": 7, "precio": 320}
    ]
)

df2
```

### Desde una lista de listas

```python
df3 = pd.DataFrame(
    [
        ["Laptop", 3, 1200],
        ["Monitor", 7, 320]
    ],
    columns=["producto", "unidades", "precio"]
)

df3
```

### Columnas calculadas

```python
df = pd.DataFrame({
    "producto": ["Laptop", "Monitor", "Teclado"],
    "unidades": [3, 7, 12],
    "precio": [1200, 320, 95]
})
df
```

```python
df["importe"] = df["unidades"] * df["precio"]
df
```

```python
df["unidades"].min()
```
