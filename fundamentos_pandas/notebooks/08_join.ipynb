```python
import pandas as pd

enero =pd.DataFrame({
    "mes":["Enero"],
    "numero_mes":[1]})
febrero =pd.DataFrame({
    "mes":["Febrero"],
    "numero_mes":[2]})
marzo=pd.DataFrame({
    "mes":["Marzo"],
    "numero_mes":[3],})
```

```python
## Concatenar por columnas
pd.concat([enero,febrero,marzo],axis=1)
```

```python
## Concatenar por filas
pd.concat([enero,febrero,marzo],axis=0)
```

```python
## Concatenar por filas y resetear indices
pd.concat([enero,febrero,marzo],axis=0).reset_index(drop=True)
```
