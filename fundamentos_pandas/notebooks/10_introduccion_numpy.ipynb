### Cast de lista a numpy array

```python
array = np.array(lista)
print(array)
print(type(array))
print(array.dtype)
```

```python
array_float = array.astype("float64")
print(array_float)
print(array_float.dtype)
```

```python
array_str = array.astype("str")
print(array_str)
print(array_str.dtype)
```

### Tipos de array

```python
# SCALAR
scalar = np.array(10)
print(scalar)
print(type(scalar))
print(scalar.ndim)
print(scalar.shape)
```

```python
# Vector
vector = np.array([1,2,3,4])
print(vector)
print(type(vector))
print(vector.ndim)
print(vector.shape)
```

```python
# MATRIZ
matriz = np.array([
    [1, 2, 3],
    [4, 5, 6]
])
print(matriz)
print(type(matriz))
print(matriz.ndim)
print(matriz.shape)
```

```python
import time

lista = list(range(10000000))

def potencia(datos, exp=2):
    return[dato ** exp for dato in datos]

start = time.time()
potencia(lista)
end= time.time()

print(f"La funcion potencia en una lista demoro: {end - start}")
```

```python
array = np.array(lista)

start = time.time()
resultado = array ** 2
end = time.time()

print(f"La funcion potencia en una array demoro: {end - start}")
```
