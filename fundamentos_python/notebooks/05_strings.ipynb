# Métodos de limpieza

```python
nombre =" AnA         TORRES    "
apellido_materno = "Zeballos"

print(f"Número de caractéres: {len(nombre)}") # len > cuenta la cantidad de caracteres
print(f"Número de caractéres: {len(apellido_materno)}")
```

```python
print(len(nombre.strip())) # .strip() > quita al inicio y al final los espacios
```

```python
print(nombre.lower()) # .lower() > convierte todo a miniscula
```

```python
print(nombre.upper()) # .upper() > convierte a todo en mayúscula
```

```python
print(nombre.title()) # .title() > pone la primera letra en mayuscula
```

```python
nombre.replace("TORRES", "LLERENA") # .replace() > remplaza el dato inicial luego la ,
```

```python
nombre.replace(" ", "").replace("AT", "A T").title()
```

```python
nombre.split() # .split() > divide donde encuentre el espacio 
```

```python
" ".join(nombre.split()).title() # .join() > para unir
```

```python
nombre = "AnaxxTorres"
", ".join(nombre.split("xx"))
```

```python
# Quitar espacios de mas con regex

import re
nombre = " AnA         TORRES    "
nombre_limpio = re.sub(r"\s+", " ", nombre).strip().title()
print(f"'{nombre_limpio}'")
```
