```python
ventas =[32,1450,890.5]
```

```python
ventas.append(2100)
ventas
```

```python
ventas.append(5000)
ventas
```

```python
ventas.append([1,2]) #.append() < agrega la lista completa [1, 2] como un único elemento nuevo al final de ventas!
ventas
```

```python
ventas.extend([4,5]) # .extend() < agrega todos los elementos de otra lista al final de la lista actual!
ventas
```

```python
ventas.insert(0,55) # .insert() < agrega el número 55 al inicio de la lista, moviendo el resto a la derecha!
ventas
```

```python
ventas.insert(2,999)
ventas 
```

```python
ventas.remove([1,2]) # .remove() < busca el primer 55 en la lista y lo elimina de inmediato!
ventas
```

```python
ventas.pop() # .pop() < elimina y devuelve el último elemento de la lista!
```

### Ordenar

```python
ventas
```

```python
ventas.sort() # .sort() < ordena de menor a mayor y modifica la lista original permanentemente!
ventas
```

```python
ventas.sort(reverse=True) # .sort(reverse=True) < ordena de mayor a menor y modifica la lista original permanentemente!
ventas
```

```python
ventas_asc = sorted(ventas) # sorted() < ordena los elementos de ventas de menor a mayor sin modificar la lista original!
ventas_asc
```

### Operaciones / agregaciones

```python
ventas = [1,2,2,2,2,3,4,5]
sum(ventas) # sum() < suma todos los números dentro de la lista ventas y devuelve el total!
```

```python
min(ventas) # min() < busca y devuelve el número con el valor más bajo dentro de ventas!
```

```python
max(ventas) # max() < busca y devuelve el número con el valor más alto dentro de ventas!
```

```python
ventas.count(2) # .count(n) < cuenta cuántas veces aparece el número 2 dentro de la lista!
```
