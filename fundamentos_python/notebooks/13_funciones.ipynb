```python
PI = 3.1416

def calc_area_circulo(radio: float, n_Decimales: int) -> float:
    return round(PI * radio**2,n_Decimales)
```

```python
calc_area_circulo(10,2)
```

```python
calc_area_circulo(3,3)
```

```python
calc_area_circulo(10)
```
