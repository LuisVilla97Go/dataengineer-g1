```python
regiones = ["Lima","Cuzco","Lima","Arequipa","Cuzco","Piura"]

set_regiones = set(regiones) # set() < crea un conjunto vacío o convierte otra estructura (como una lista) en un conjunto sin duplicados!

print(f"lista regiones: {regiones}")
print(f"conjuto (set) regiones: {set_regiones}")
```

```python
len(regiones)
```

```python
len(set_regiones)
```

```python
sedes_esperadas={"Arequipa", "Piura","Loreto"}
```

### Union ("|" - or)

```python
set_regiones | sedes_esperadas
```

```python
set_regiones.union(sedes_esperadas) # .union() < ombina los elementos de ambos conjuntos en uno nuevo, eliminando los duplicados!
```

### Interseccion ("&" - and)

```python
set_regiones & sedes_esperadas
```

```python
set_regiones.intersection(sedes_esperadas) # .intersection() < devuelve un nuevo conjunto con los elementos que están en AMBOS conjuntos!
```

### Diferencia("-")

```python
sedes_esperadas - sedes_esperadas
```

```python
sedes_esperadas - set_regiones
```

```python
sedes_esperadas.difference(set_regiones) # .difference() < devuelve los elementos del primer conjunto que NO están en el segundo!
```

### Diferencia simétrica

```python
set_regiones ^ sedes_esperadas
```

```python
set_regiones.symmetric_difference(sedes_esperadas) # .symmetric_difference() < devuelve los elementos que están en uno u otro conjunto, pero NO en ambos!
```

### Subconjutos

```python
sedes_esperadas
```

```python
region_elegida = {"Cuzco"}
```

```python
set_regiones.issubset(sedes_esperadas) # .issubset() < verifica si todos los elementos de un conjunto están dentro de otro!
```

```python
sedes_esperadas.issubset(set_regiones) 
```

```python
set_regiones.issubset(region_elegida)
```

```python
region_elegida.issubset(set_regiones)
```

```python
set_regiones.issuperset(region_elegida) # .issuperset() < verifica si un conjunto contiene a todos los elementos de otro conjunto!
```
