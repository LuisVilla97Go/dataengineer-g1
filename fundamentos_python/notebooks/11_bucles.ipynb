```python
columnas = ["id", "region", "cantidad"]
valores = ["V-101", "Lima", "2"]
```

```python
### NOES PYTHONISTA, NO ES UNA BUENA PRÁCTICA DE CÓDIGO LIMPIO
for idx in range(len(columnas)):
    print(f"{columnas[idx]}: {valores[idx]}")
```

```python
for columna, valor in zip(columnas, valores):
    print(f"{columna}: {valores}")
```

### BUCLE: WHILE

```python
intentos = 0

while intentos < 3:
    print(f"Intento: {intentos}")
    intentos += 1
```

```python
numero = int(input("Ingrese número: "))

while numero < 10:
    print(f"El número es: {numero}")
    numero = int(input({"ingresa número"}))

print(f"El número ganador es: {numero}")
```

### Control o flujo de ejecución

```python
for numero in range(20):
    if numero % 2 == 0:
        continue

    print(f"Número impar: {numero}")
```

```python
for numero in range(20):
    if numero % 2 == 0:
        continue

    if numero >= 10:
        break
    
    print(f"Número impar: {numero}")

print("El bucle finalizó")
```
