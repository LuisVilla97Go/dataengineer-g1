### Expresiones condicionales

```python
registros = []

print("Llenar los registros de hoy dia...")

if registros:
    print("Procesando...")
else:
    print("Los registros están vacíos")
```

```python
registros = []

print("Llenar los registros de hoy dia...")
registros.extend([1,2,3,4,5])

if registros:
    print("Procesando...")
else:
    print("Los registros están vacíos")
```

```python
### NADA PYTHONISTA, NO SE DEBERIA PARA TENER UN CODIGO MAS LIMPIO

cantidad_registros = len(registros)

if cantidad_registros > 0:
    print("Procesando...")
```
