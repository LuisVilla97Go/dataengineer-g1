### Pregunta 3.1: Consolidación de inventarios

Combina tres inventarios provenientes de distintas sucursales. Cuando un producto aparezca en varias
sucursales, **suma sus cantidades** y calcula también **en cuántas sucursales** está disponible.

**Requisitos:**
- Crea un diccionario final donde cada producto tenga las claves `cantidad_total` y `sucursales`.
- No modifiques los diccionarios originales.
- Ordena el resultado alfabéticamente por producto.

```python
inventario_norte = {"laptop": 4, "mouse": 10, "monitor": 3}
inventario_sur = {"laptop": 2, "teclado": 8, "monitor": 5}
inventario_centro = {"mouse": 6, "teclado": 4, "webcam": 7}
```

```python
# Escribe tu solución aquí

inventario_total = {}

inventarios = [inventario_norte, inventario_sur, inventario_centro]

articulos = set([
    *inventario_norte.keys(),
    *inventario_sur.keys(),
    *inventario_centro.keys(),
])

print(articulos)
print("="*50)

for articulo in sorted(articulos):
    inventario_total[articulo] = {
        "cantidad_total": sum([inventario.get(articulo, 0) for inventario in inventarios]),
        "sucursales": sum([1 for inventario in inventarios if articulo in inventario])
    }

for articulo in inventario_total:
    print(f"articulo: {articulo} | {inventario_total[articulo]}")
    print('='*50)
```

### Pregunta 3.3: Agrupación jerárquica de empleados

Agrupa a los empleados primero por **departamento** y luego por **nivel** de experiencia.

**Requisitos:**
- En cada grupo almacena `nombres`, `cantidad` y `salario_promedio`.
- La estructura debe ser un diccionario anidado: `{departamento: {nivel: {...}}}`.

```python
empleados = [
    {"nombre": "Ana", "departamento": "Desarrollo", "nivel": "Senior", "salario": 5200},
    {"nombre": "Luis", "departamento": "Diseño", "nivel": "Junior", "salario": 2800},
    {"nombre": "Carla", "departamento": "Desarrollo", "nivel": "Junior", "salario": 3400},
    {"nombre": "Pedro", "departamento": "Desarrollo", "nivel": "Senior", "salario": 5600},
    {"nombre": "Lucía", "departamento": "Diseño", "nivel": "Senior", "salario": 4700},
]
```

```python
# Escribe tu solución aquí
agrupados = {}

for empleado in empleados:
    departamento = empleado["departamento"]
    nivel = empleado["nivel"]

    agrupados.setdefault(departamento, {})
    agrupados[departamento].setdefault(nivel, [])
    agrupados[departamento][nivel].append(empleado)


for departamento in agrupados:
    for nivel in agrupados[departamento]:
        grupo = agrupados[departamento][nivel]
        salarios = [empleado["salario"] for empleado in grupo]

        agrupados[departamento][nivel] = {
            "nombres": [empleado["nombre"] for empleado in grupo],
            "cantidad": len(grupo),
            "salario_promedio": sum(salarios)/len(salarios)
        }

for departamento, niveles in agrupados.items():
    print(departamento)
    for nivel, resumen in niveles.items():
        print(f"    {nivel}: {resumen}")
```

### Pregunta 3.4: Sistema de votación con validación

Procesa una lista de votos y separa los votos **válidos** de los **inválidos**.

**Requisitos:**
- Los votos deben ser insensibles a mayúsculas (`"ANA"` cuenta para `"Ana"`).
- Registra el conteo por candidato, los votos inválidos y el porcentaje obtenido (sobre los votos válidos).
- Determina si hay un ganador o un empate.

```python
candidatos = {"Ana", "Luis", "Carla"}
votos = ["Ana", "Luis", "ana", "Carla", "Pedro", "Luis", "ANA", "", "Carla"]
```

```python
# Escribe tu solución aquí
conteo = {}
for candidato in sorted(candidatos):
    conteo[candidato] = 0

votos_invalidos = []

for voto in votos:
    voto_normalizado = voto.strip().title()
    if voto_normalizado in candidatos:
        conteo[voto_normalizado] += 1
    else:
        votos_invalidos.append(voto)

total_validos = sum(conteo.values())

porcentajes = {}
for candidato, cantidad in conteo.items():
    if total_validos > 0:
        porcentajes[candidato] = round(cantidad /total_validos * 100, 2)
    else:
        porcentajes[candidato] = 0

print(f"Conteo: {conteo}")
print(f"Porcentajes: {porcentajes}")
print(f"Vatos válidos: {total_validos}")
print(f"Vatos inválidos: {len(votos_invalidos)} -> {votos_invalidos}")
```

```python
mayor_votacion = max(conteo.values())
lideres = [candidato for candidato, cantidad in conteo.items() if cantidad == mayor_votacion]

if mayor_votacion == 0:
    print("No hubo votos válidos")
elif len(lideres) == 1:
    print(f"Ganador: {lideres[0]} con {mayor_votacion} votos")
else:
    print(f"Empate entre {lideres} con {mayor_votacion} votos cada uno")
```

### Pregunta 3.5: Índice invertido de documentos

Crea un **índice invertido** que permita saber en qué documentos aparece cada palabra.

- Ignora las palabras de una sola letra.

```python
documentos = {
    "doc1": "python y ciencia de datos",
    "doc2": "python para desarrollo web",
    "doc3": "ciencia de datos con inteligencia artificial",
}
```

```python
# Escribe tu solución aquí
indice = {}

for doc_id, texto in documentos.items():
    for palabra in texto.split():
        if len(palabra) > 1:
            indice.setdefault(palabra, set())
            indice[palabra].add(doc_id)

for palabra, docs in indice.items():
    print(f"{palabra}: {docs}")
```

### Pregunta 5.1: Analizar las ventas de una tienda

Crea una función `analizar_ventas(ventas)` que reciba una lista de montos de venta y construya un resumen estadístico.

**Requisitos:**
- Calcular el total vendido, el promedio, la venta más alta y la venta más baja.
- Contar cuántas ventas fueron superiores al promedio.
- Indicar cuántas ventas se encuentran por debajo de 100.
- Retornar todos los resultados dentro de un diccionario.
- Si la lista está vacía o contiene valores negativos, retornar un mensaje de error.

**Reto adicional:** agrega un parámetro opcional `meta`. La función debe indicar si el total vendido alcanzó dicha meta.

```python
ventas = [150.50, 200, 80.90, 300, 175.50, 90]
```

```python
# Escribe tu solución aquí
def analizar_ventas(ventas: list[float]) -> dict[str, float] | str:
    if not ventas:
        return "ERROR: No existen ventas"

    for venta in ventas:
        if venta < 0:
            return f"ERROR: Valor de venta negativo: {venta}"

    total_vendido = sum(ventas)
    promedio_vendido = total_vendido / len(ventas)
    venta_max = max(ventas)
    venta_min = min(ventas)

    ventas_superiores_promedio = sum([1 for venta in ventas if venta >= promedio_vendido])
    ventas_por_debajo_100 = sum([1 for venta in ventas if venta < 100])

    estadisticos = {
        "total_vendido": total_vendido,
        "promedio_vendido": promedio_vendido,
        "venta_max": venta_max,
        "venta_min": venta_min,
        "ventas_superiores": ventas_superiores_promedio,
        "ventas_por_debajo_100": ventas_por_debajo_100
    }

    return estadisticos


stats = analizar_ventas(ventas)
print(stats)
```

```python
stats = analizar_ventas([])
print(stats)
```

```python
stats = analizar_ventas([150.50, 200, 80.90, 300, 175.50, 90, -100])
print(stats)
```

```python
def analizar_ventas_meta(venta: list[float], meta: float | None = None) -> dict[str, float] | str:
    estadisticos = analizar_ventas(ventas)

    if type(estadisticos) == str or meta is None:
        return estadisticos

    estadisticos["meta"] = meta
    estadisticos["meta_alcanzada"] = estadisticos["total_vendido"] >= meta
    return estadisticos

stats = analizar_ventas_meta(ventas, meta=1000)
print(stats)
```

### Pregunta 5.3: Evaluar el rendimiento de empleados

Crea una función `evaluar_empleados(empleados, minimo=70)` que analice las evaluaciones de cada empleado.

**Requisitos:**
- Cada empleado tendrá una lista de puntajes entre 0 y 100.
- Calcular el promedio de cada empleado.
- Clasificar el resultado como **Excelente** (≥ 90), **Bueno** (≥ 80), **Regular** (≥ 70) o **Deficiente** (< 70).
- Incluir únicamente a quienes tengan un promedio igual o superior a `minimo`.
- Retornar una lista de tuplas con `(nombre, promedio, clasificación)`.
- Ordenar de mayor a menor promedio y, en caso de empate, alfabéticamente.

```python
empleados = {
    'Ana': [80, 90, 75],
    'Carlos': [60, 70, 65],
    'Lucía': [95, 85, 90],
    'Pedro': [70, 70, 70],
}
```

```python
# Escribe tu solución aquí
def clasificar(promedio: float) -> str:
    if promedio >= 90:
        return "Excelente"
    elif promedio >= 80:
        return "Bueno"
    elif promedio >= 70:
        return "Regular"
    else:
        return "Deficiente"

def criterio_orden(empleado: tuple[str, float, str]) -> tuple[float, str]:
    return -empleado[1], empleado[0]

def evaluar_empleados(empleados: dict[str, list[int]], minimo: float = 70) -> list[tuple[str, float, str]] | str:
    resultado = []

    for nombre, puntajes in empleados.items():
        if not puntajes:
            return f"ERROR: {nombre} no tiene puntajes"

        for puntaje in puntajes:
            if puntaje < 0 or puntaje > 100:
                return f"ERROR: Puntaje fuera de rango para {nombre}: {puntaje}"

        promedio = round(sum(puntajes) / len(puntajes), 2)

        if promedio >= minimo:
            resultado.append((nombre, promedio, clasificar(promedio)))

    resultado.sort(key=criterio_orden)
    return resultado


evaluar_empleados(empleados)
```

### Pregunta 5.4: Comprimir y descomprimir texto

Implementa dos funciones: `comprimir_texto(texto)` y `descomprimir_texto(texto)`.

**Requisitos:**
- `comprimir_texto` debe reemplazar caracteres consecutivos repetidos por el carácter y su cantidad.
  Por ejemplo, `aaabbccccdaa` debe convertirse en `a3b2c4d1a2`.
- `descomprimir_texto` debe reconstruir el texto original.
- La descompresión debe admitir cantidades de más de un dígito, como `a12`.
- Si el formato comprimido es inválido, retornar un mensaje de error.

```python
texto = 'aaabbccccdaa'
```

```python
# Escribe tu solución aquí
def comprimir_texto(texto: str) -> str:
    if not texto:
        return ""
    
    resultado = ""
    actual = texto[0]
    contador = 1

    for caracter in texto[1:]:
        if caracter == actual:
            contador += 1
        else:
            resultado += f"{actual}{contador}"
            actual = caracter
            contador = 1

    resultado += f"{actual}{contador}"
    return resultado

def descomprimir_texto(texto: str) -> str:
    resultado = ""
    i = 0

    while i < len(texto):
        caracter = texto[i]
        if caracter.isdigit():
            return f"ERROR: Formato inválido, se esperaba una letra en la posición {i}"
        i += 1

        cantidad = ""
        while i < len(texto) and texto[i].isdigit():
            cantidad += texto[i]
            i += 1

        if cantidad == "":
            return f"ERROR: Formato inválido, falta la cantidad después de '{caracter}'"
        if int(cantidad) == 0:
            return f"ERROR: Formato inválido, la cantidad de '{caracter}' no puede ser 0"

        resultado += caracter * int(cantidad)

    return resultado

comprimido = comprimir_texto(texto)
print(f"Comprimido: {comprimido}")
print(f"Descomprimido: {descomprimir_texto(comprimido)}")
```

### Pregunta 5.5: Sistema de reservas de asientos

Crea una función `reservar_asientos(sala, asientos_solicitados)` que permita reservar varios asientos de una sala.

**Requisitos:**
- La sala será una lista de listas donde `L` significa libre y `O` significa ocupado.
- Cada asiento solicitado se expresará mediante una tupla `(fila, columna)`, **numeradas desde 1**.
- Validar que todas las posiciones existan y que los asientos estén libres.
- Realizar la reserva únicamente si todos están disponibles.
- Evitar reservas parciales y mantener la sala intacta si ocurre un error.
- Retornar el estado, los asientos reservados y la sala actualizada.

**Reto adicional:** crea otra función `mostrar_sala(sala)` que muestre filas y columnas de manera legible.

```python
sala = [['L', 'L', 'O', 'L'],
        ['O', 'L', 'L', 'L'],
        ['L', 'O', 'L', 'L']]
asientos = [(1, 2), (2, 3), (3, 1)]
```

```python
# Escribe tu solución aquí
def mostrar_sala(sala: list[list[str]]) -> None:
    if not sala:
        print("La sala está vacía")
        return

    columnas = [f"C{numero}" for numero in range(1, len(sala[0]) + 1)]
    print("        " + "  ".join(columnas))

    for numero, fila in enumerate(sala):
        print(f"Fila {numero + 1}:  " + "   ".join(fila))

def reservar_asientos(sala: list[list[str]], asientos_solicitados: list[tuple[int, int]]) -> dict:
    errores = []

    if len(set(asientos_solicitados)) != len(asientos_solicitados):
        errores.append("Hay asientos repetidos en la solicitud")

    for fila, columna in asientos_solicitados:
        if fila < 1 or fila > len(sala) or columna < 1 or columna > len(sala[fila - 1]):
            errores.append(f"El asiento ({fila}, {columna}) no existe")
        elif sala[fila - 1][columna - 1] == "O":
            errores.append(f"El asiento ({fila}, {columna}) ya está ocupado")

    if errores:
        return {"estado": "ERROR", "errores": errores, "reservados": [], "sala": sala}

    sala_actualizada = [list(fila) for fila in sala]

    for fila, columna in asientos_solicitados:
        sala_actualizada[fila - 1][columna - 1] = "O"

    return {"estado": "OK", "errores": [], "reservados": asientos_solicitados, "sala": sala_actualizada}

mostrar_sala(sala)
```

```python
resultado = reservar_asientos(sala, asientos)

print(f"Estado: {resultado['estado']}")
print(f"Reservados: {resultado['reservados']}")
sala = resultado["sala"]
mostrar_sala(sala)
```

```python
resultado = reservar_asientos(sala, [(1, 4), (1, 2), (4, 1)])

print(f"Estado: {resultado['estado']}")
print(f"Errores: {resultado['errores']}")
mostrar_sala(resultado["sala"])
```

### Pregunta 6.1: Clase CuentaBancaria

Crea una clase `CuentaBancaria` con los atributos `titular` y `saldo` (el saldo inicial por defecto será 0).

- Método `depositar(monto)`: suma al saldo, solo si el monto es positivo.
- Método `retirar(monto)`: resta del saldo, pero debe impedir que el saldo quede negativo y avisar `"Fondos insuficientes"`.
- Método `consultar_saldo()`: devuelve el saldo actual.

Prueba la clase con una cuenta: deposita, retira un monto válido y luego intenta retirar más de lo disponible.

*Objetivo: modificar el estado interno del objeto con validaciones.*

```python
# No hay datos de entrada: crea tus propios objetos para probar la clase.
```

```python
# Escribe tu solución aquí
class CuentaBancaria:
    def __init__(self, titular: str, saldo: float = 0) -> None:
        self.titular = titular
        self._saldo = saldo

    def _monto_positivo(self, monto: float) -> bool:
        return monto > 0

    def depositar(self, monto: float) -> None:
        if not self._monto_positivo(monto):
            print(f"ERROR, monto no permitido. Monto={monto}")
        else:
            self._saldo += monto

    def retirar(self, monto: float) -> None:
        if not self._monto_positivo(monto):
            print(f"ERROR, monto no permitido. Monto={monto}")
        elif monto > self._saldo:
            print("Fondos insuficientes")
        else:
            self._saldo -= monto

    def consultar_saldo(self) -> float:
        return self._saldo
```

```python
# Escribe tu solución aquí
class Motor:
    def __init__(self, tipo: str, caballos: int) -> None:
        self.tipo = tipo
        self.caballos = caballos
        self.encendido = False

    def arrancar(self) -> str:
        if self.encendido:
            return "El motor ya está encendido"
        self.encendido = True
        return f"Motor {self.tipo} de {self.caballos} HP encendido"

class Vehiculo:
    def __init__(self, marca: str, motor: Motor) -> None:
        self.marca = marca
        self.motor = motor

    def arrancar(self) -> str:
        return f"{self.marca}: {self.motor.arrancar()}"
```

### Pregunta 6.5: Sistema de biblioteca (ejercicio integrador)

Crea tres clases que trabajen juntas: `Libro`, `Usuario` y `Biblioteca`.

- `Libro`: `titulo`, `autor` y `disponible` (`True` por defecto). *(Es una clase distinta a la del ejercicio 6.2: puedes redefinirla.)*
- `Usuario`: `nombre` y una lista de `libros_prestados`.
- `Biblioteca`: guarda una lista de libros y ofrece los métodos `agregar_libro()`, `buscar(titulo)`,
  `prestar(titulo, usuario)`, `devolver(titulo, usuario)` y `catalogo()`.
- Al prestar un libro debe marcarse como no disponible y añadirse a la lista del usuario; al devolverlo, lo contrario.
- Contempla los casos: el libro no existe, el libro ya está prestado, el usuario no tiene ese libro.

*Objetivo: integrar varias clases que se relacionan entre sí, tal como ocurre en un programa real.*

```python
# No hay datos de entrada: crea tus propios objetos para probar las clases.
```

```python
# Escribe tu solución aquí
class Libro:
    def __init__(self, titulo: str, autor: str, disponible: bool = True) -> None:
        self.titulo = titulo
        self.autor = autor
        self.disponible = disponible

    def mostrar_info(self) -> str:
        estado = "Disponible" if self.disponible else "Prestado"
        return f"{self.titulo} - {self.autor} [{estado}]"

class Usuario:
    def __init__(self, nombre: str) -> None:
        self.nombre = nombre
        self.libros_prestados = []


class Biblioteca:
    def __init__(self) -> None:
        self.libros = []

    def agregar_libro(self, libro: Libro) -> None:
        self.libros.append(libro)

    def buscar(self, titulo: str) -> Libro | None:
        for libro in self.libros:
            if libro.titulo.lower() == titulo.lower():
                return libro
        return None

    def prestar(self, titulo: str, usuario: Usuario) -> str:
        libro = self.buscar(titulo)
        if libro is None:
            return f"ERROR: El libro '{titulo}' no existe"
        if not libro.disponible:
            return f"ERROR: El libro '{titulo}' ya está prestado"

        libro.disponible = False
        usuario.libros_prestados.append(libro)
        return f"'{libro.titulo}' prestado a {usuario.nombre}"

    def devolver(self, titulo: str, usuario: Usuario) -> str:
        libro = self.buscar(titulo)
        if libro is None:
            return f"ERROR: El libro '{titulo}' no existe"
        if libro not in usuario.libros_prestados:
            return f"ERROR: {usuario.nombre} no tiene el libro '{titulo}'"

        libro.disponible = True
        usuario.libros_prestados.remove(libro)
        return f"{usuario.nombre} devolvió '{libro.titulo}'"

    def catalogo(self) -> None:
        for libro in self.libros:
            print(libro.mostrar_info())
```

```python
biblioteca = Biblioteca()
biblioteca.agregar_libro(Libro("Habitos Atomicos", "James Clear"))
biblioteca.agregar_libro(Libro("Deja de ser tu", "Joe Dispenza"))
biblioteca.agregar_libro(Libro("Rompe la barrera del No", "Chris Boss"))

ana = Usuario("Ana")
luis = Usuario("Luis")

biblioteca.catalogo()
```
